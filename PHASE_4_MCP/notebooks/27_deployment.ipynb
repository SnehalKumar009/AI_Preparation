# 27 · Deployment & config (`mcp.json`)

Hosts (editors, desktop apps) load servers from a small JSON config: a name,
the launch command, args, and environment. The same servers you built run
unchanged under any MCP-compatible host.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import json, sys
config = {'mcpServers': {
  'calculator': {'command': sys.executable, 'args': ['-m','mcp_lab.servers.calculator']},
  'filesystem': {'command': sys.executable, 'args': ['-m','mcp_lab.servers.filesystem'],
                 'env': {'MCP_FS_ROOT': '.'}},
  'code_search': {'command': sys.executable, 'args': ['-m','mcp_lab.servers.code_search'],
                  'env': {'MCP_CODE_ROOT': '.'}}}}
print(json.dumps(config, indent=2))
```

Drop an equivalent block into your host's MCP settings to expose these
servers to it. Env vars set each server's sandbox root.
