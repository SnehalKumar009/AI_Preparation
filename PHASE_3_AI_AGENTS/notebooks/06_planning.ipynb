# 06 · Planning — Task Decomposition

Complex tasks are easier when broken into steps. **Decomposition** turns a fuzzy goal into an ordered list of concrete sub-tasks; **Plan-and-Execute** then runs each sub-task with a tool-using agent and composes the results.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
```

```python
from agents_lab import decompose

task = 'Plan a 3-day trip to Kyoto: pick neighborhoods, list must-see sites, and a food plan.'
steps = decompose(task, provider='ollama')
for i, s in enumerate(steps, 1):
    print(f'{i}. {s}')
```

**Plan-and-Execute** on a task that needs tools: it plans, runs each step with an `Agent` (calculator available), then composes a final answer. The shared `CostTracker` totals every call.

```python
from agents_lab import PlanAndExecute

pae = PlanAndExecute(provider='ollama')
res = pae.run('A recipe needs 3 eggs per cake. Compute eggs for 4 cakes, then for 7 cakes, then the total.')
print('ANSWER:\n', res.answer)
print('\ncost:', pae.tracker.summary())
```

**Takeaway:** planning separates *deciding what to do* from *doing it*. Decompose first, execute per step, then compose.
