# 01 · Embeddings

**Idea:** an embedding maps text to a vector so *similar meanings sit close together* — the foundation of every retriever in this phase. We compare two clients: **Ollama** `nomic-embed-text` (local) and **OpenAI** `text-embedding-3-small` (cloud).

> `rag_lab` auto-finds Phase 1 `study_buddy`, so you only add `RAG_ENGINEERING` to the path.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

import numpy as np
from rag_lab import embed_texts, cosine
```

```python
texts = [
    'A dog barked loudly at the mailman.',
    'The puppy made noise at the delivery person.',
    'Quarterly revenue grew by twelve percent.',
]
vecs = embed_texts(texts, provider='ollama')  # L2-normalized by default
print('ollama shape:', vecs.shape)
```

Same text, a different embedding model. Dimensions differ — embeddings are only comparable *within* the same model. This cell skips cleanly if no OpenAI key is set.

```python
try:
    ovecs = embed_texts(texts, provider='openai')
    print('openai shape:', ovecs.shape)
except Exception as e:
    print('openai skipped:', type(e).__name__, '-', e)
```

```python
# Meaning becomes geometry: related texts score high, unrelated low.
print('dog vs puppy  :', round(cosine(vecs[0], vecs[1]), 3), '(similar)')
print('dog vs revenue:', round(cosine(vecs[0], vecs[2]), 3), '(unrelated)')
```

**Takeaway:** embeddings turn meaning into geometry; cosine similarity ranks relevance. Vectors are only comparable within one model, so pick an embedding model and stick to it.
