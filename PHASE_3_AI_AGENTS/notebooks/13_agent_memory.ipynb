# 13 · Agent Short-Term Memory

Agents need somewhere to keep what they've learned *within* a task: a running **scratchpad** of notes, and a **rolling conversation window** that stays under a token budget. This is the short-term working memory the loop uses — the full taxonomy (long-term, semantic, episodic, procedural) is Phase 6.

> Runs locally; no model calls needed for the core demo.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

A **scratchpad** accumulates notes the agent can render back into a prompt.

```python
from agents_lab import Scratchpad

pad = Scratchpad()
pad.add('User is planning a trip to Kyoto in November.')
pad.add('Budget is mid-range; prefers walking.')
print(pad.render())
print('notes:', len(pad))
```

**Short-term memory** keeps a rolling window under a token cap — old turns are dropped so context never overflows.

```python
from agents_lab import ShortTermMemory

mem = ShortTermMemory(max_tokens=60)
for i in range(8):
    mem.add('user', f'This is message number {i} with some filler content to use tokens.')
print('messages kept:', len(mem.window()))
for m in mem.window():
    print(' ', m['content'][:50])
```

**Takeaway:** short-term memory keeps an agent coherent within a task while staying inside the context window. Persistent, cross-session memory is Phase 6.
