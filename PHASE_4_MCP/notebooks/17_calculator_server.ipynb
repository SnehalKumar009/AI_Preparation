# 17 · Calculator server (safe eval)

The smallest server: one `evaluate` tool using a restricted `ast` evaluator —
never Python `eval` — so a hostile expression cannot run arbitrary code.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.calculator']
```

```python
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print('2*(3+4)', c.call_tool('evaluate', {'expression': '2*(3+4)'}))
    print('2**10', c.call_tool('evaluate', {'expression': '2**10'}))
```
