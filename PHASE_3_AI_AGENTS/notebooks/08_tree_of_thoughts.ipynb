# 08 · Tree of Thoughts

Chain-of-Thought commits to one line of reasoning. **Tree of Thoughts (ToT)** explores several: at each level it *expands* candidate next-thoughts, *scores* how promising each is, keeps the best few (beam search), and continues to a set depth. More model calls, but far better on problems where a greedy path goes wrong.

> Runs on Ollama. This makes many calls — keep `breadth` and `depth` small.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
```

```python
from agents_lab import tree_of_thoughts
from study_buddy import CostTracker

problem = 'Use the numbers 4, 9, 10, 13 with + - * / to make 24. Show the expression.'
tracker = CostTracker()
answer, beam = tree_of_thoughts(problem, provider='ollama', breadth=3, depth=2, beam=2, tracker=tracker)
print('ANSWER:\n', answer)
print('\ncost:', tracker.summary())
```

Inspect the surviving beam — the highest-scored reasoning paths ToT kept at the final level.

```python
for node in beam:
    print(f'score={node.score:.1f}  path={node.path}')
```

**Takeaway:** ToT = search over reasoning. Expand → score → keep the best. Powerful for puzzles and planning, but budget the calls (see notebook 12).
