# 11 · Structured Output (JSON)

**Idea:** apps need machine-readable output, not prose. Ask the model to return JSON matching a schema, then parse it into a Python object. `study_buddy` passes `response_format` through to each provider.

```python
import sys, pathlib, json
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')  # Ollama accepts a JSON schema via format=
```

```python
schema = {
    'type': 'object',
    'properties': {
        'title': {'type': 'string'},
        'ingredients': {'type': 'array', 'items': {'type': 'string'}},
        'minutes': {'type': 'integer'},
    },
    'required': ['title', 'ingredients', 'minutes'],
}

messages = [
    {'role': 'system', 'content': 'Extract a recipe as JSON. Return ONLY JSON.'},
    {'role': 'user', 'content': 'A 15-minute garlic butter pasta with spaghetti, garlic, butter, parsley.'},
]
resp = llm.chat(messages, temperature=0, response_format={'schema': schema})
print(resp.text)
```

```python
data = json.loads(resp.text)
print('title      :', data['title'])
print('minutes    :', data['minutes'], '(type', type(data['minutes']).__name__ + ')')
print('ingredients:', ', '.join(data['ingredients']))
```

**Takeaway:** structured output makes LLMs programmable components. Combined with tool calling, it's the backbone of reliable agents.

> For OpenAI, pass `response_format={'type': 'json_object'}` (or a json_schema). Gemini sets `response_mime_type` automatically here.
