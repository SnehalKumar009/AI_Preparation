# 19 · Framework Bridge — Same Ideas, Different APIs

Everything so far was built by hand so the mechanics are visible. Production code usually leans on a framework. This notebook maps our `ToolRegistry` onto two popular ones — the **OpenAI Agents SDK** and **LangGraph** — so you see they're the *same ideas* (tools + a loop) with different APIs. Deep framework coverage (LangGraph, Temporal, CrewAI, AutoGen, Semantic Kernel, ADK) is **Phase 5**.

> Optional dependencies. Cells skip cleanly if a framework isn't installed.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

Our from-scratch tools convert straight into an **OpenAI Agents SDK** agent (`pip install openai-agents`).

```python
from agents_lab import default_tools, frameworks

try:
    oa_agent = frameworks.to_openai_agents(default_tools(), instructions='You are a helpful math agent.')
    print('built OpenAI Agents SDK agent:', oa_agent)
    # from agents import Runner; print(Runner.run_sync(oa_agent, 'What is 6*7?').final_output)
except ImportError as e:
    print('openai-agents not installed:', e)
```

The same tools also drop into a **LangGraph** prebuilt ReAct graph (`pip install langgraph langchain`).

```python
try:
    graph = frameworks.to_langgraph(default_tools(), provider_model='openai:gpt-4o-mini')
    print('built LangGraph agent:', type(graph).__name__)
    # print(graph.invoke({'messages': [('user', 'What is 6*7?')]}))
except ImportError as e:
    print('langgraph not installed:', e)
```

**Takeaway:** frameworks don't add new *concepts* — they package the loop, tools, and state you already understand. Because you built it from scratch, you can pick any framework and know exactly what it's doing. Phase 5 goes deep on orchestration frameworks.
