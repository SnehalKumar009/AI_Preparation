# 22 · Streaming an Agent

Users should not stare at a frozen screen while an agent works. Two levels of
streaming:

1. **Event streaming** — surface each step (thought / action / observation)
   the moment it happens, via the `on_step` hook on `Agent`.
2. **Token streaming** — stream the characters of a single LLM reply, via the
   provider's `stream()` (the Phase 1 primitive).

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

### 1) Event streaming — watch the loop live
`on_step` fires as each `Step` is recorded, so the trajectory prints in real
time instead of only at the end.

```python
from agents_lab import Agent, default_tools

def show(step):
    print(f'  [{step.kind:<11}] {step.content[:80]}')

agent = Agent(provider='ollama', tools=default_tools(), on_step=show)
result = agent.run('What is 123 * 7? Use the calculator, then explain.')
print('\nFINAL:', result.answer)
```

### 2) Token streaming — stream a single reply
For the final synthesis you can stream characters as they arrive.

```python
from study_buddy import get_provider
llm = get_provider('ollama')
for chunk in llm.stream([{'role':'user','content':'In one sentence, what is an agent?'}]):
    print(chunk, end='', flush=True)
print()
```

**Takeaway:** event streaming shows *what the agent is doing*; token streaming
shows *what it is saying*. Real agent UIs use both.
