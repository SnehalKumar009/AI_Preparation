# 04 · Overlap

**Idea:** an idea that straddles a chunk boundary can be lost from *both* chunks. Overlap repeats a slice between neighbors so the idea survives whole in at least one. We show a boundary fact becoming retrievable once overlap is added.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

from rag_lab import fixed_chunks, NumpyStore
# A fact deliberately placed so it spans a chunk boundary.
doc = ('padding ' * 55) + 'The activation code is ZEPHYR-7788 issued in Berlin. ' + ('filler ' * 55)
query = 'What is the activation code issued in Berlin?'
```

```python
for overlap in [0, 25]:
    store = NumpyStore()
    store.add(fixed_chunks(doc, chunk_size=60, overlap=overlap))
    hit = store.search(query, k=1)[0]
    has = 'ZEPHYR-7788' in hit.chunk.text
    print(f'overlap={overlap:>2} | top score={hit.score:.3f} | code present: {has}')
```

With zero overlap the code can be split across two chunks and neither ranks well. A modest overlap keeps the whole fact in one chunk, so retrieval recovers it.

**Takeaway:** overlap (≈10–20% of chunk size) protects boundary-spanning facts. Too much overlap inflates the index and returns near-duplicates.
