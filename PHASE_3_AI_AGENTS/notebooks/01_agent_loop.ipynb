# 01 · The Agent Loop

An **agent** is an LLM that can *think, decide, and act* by calling tools. The engine that makes this possible is the **agent loop**:

> perceive the task → **think** → **act** (call a tool) → **observe** the result → repeat → answer.

This notebook runs the loop end-to-end with `agents_lab.Agent`, which uses the provider's **native tool calling** and feeds every tool result back until the model produces a final answer.

> Runs on Ollama. Cloud cells skip cleanly if a key is missing.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

The default toolset is a calculator + a clock (reused from Phase 1). Give the agent a task it can't answer without a tool and watch it loop.

```python
from agents_lab import Agent, default_tools

agent = Agent(provider='ollama', tools=default_tools(), max_steps=5)
result = agent.run('What is 23 * 17, and what is that plus 100? Use the calculator.')
print('ANSWER:', result.answer)
print('tool calls:', result.n_tool_calls)
```

The whole trajectory is captured. Print it to *see* the think → act → observe cycle the loop went through.

```python
from agents_lab import Tracer

tracer = Tracer()
res = agent.run('What time is it right now? Then multiply 6 by 7.', tracer=tracer)
print(tracer.pretty())
print('\ncost:', res.tracker.summary())
```

**Takeaway:** the agent loop is just *call the model → run any tools it asked for → give it the results → repeat*. Everything else in Phase 3 (ReAct, planning, reflection) is a smarter way to run this same loop.
