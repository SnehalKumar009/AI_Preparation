# 14 · Hallucination

**Idea:** LLMs confidently invent facts they don't know. The fix is *grounding*: give the model the source text and tell it to answer only from that. We first provoke a hallucination, then fix it with retrieved context (mini-RAG).

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')
```

```python
# Ask about a fake product -> the model may invent details.
q = 'What are the specs of the Zephyr X9 quantum laptop released in 2024?'
print('UNGROUNDED:\n', llm.chat([{'role': 'user', 'content': q}], temperature=0).text[:400])
```

```python
# Grounded: provide the only allowed source + instruct to say 'I don't know'.
context = 'Product catalog: We currently sell the Aero 13 and Aero 15 laptops only.'
grounded = [
    {'role': 'system', 'content': 'Answer ONLY from the provided context. If it is not there, say "I don\'t know".'},
    {'role': 'user', 'content': f'Context:\n{context}\n\nQuestion: {q}'},
]
print('GROUNDED:\n', llm.chat(grounded, temperature=0).text)
```

**Takeaway:** grounding + an explicit 'say I don't know' instruction dramatically cuts hallucination. Full RAG (next) automates the retrieval of that context.
