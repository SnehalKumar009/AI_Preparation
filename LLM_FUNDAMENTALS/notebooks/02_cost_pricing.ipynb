# 02 · Cost & Pricing (real USD)

**Idea:** cloud LLMs bill per token (input and output priced separately). Local Ollama is free but still costs *time*. Here we run the same prompt on multiple providers and compare **real USD** via `CostTracker`.

> Requires `.env` keys for the cloud providers. Ollama-only? Just run the first provider.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider, CostTracker

prompt = [{'role': 'user', 'content': 'In one sentence, what is a transformer in ML?'}]
tracker = CostTracker()
```

```python
# Add/remove providers depending on which keys you have in .env.
for name in ['ollama', 'openai', 'gemini', 'deepseek', 'anthropic']:
    try:
        resp = get_provider(name).chat(prompt, temperature=0)
        rec = tracker.add(resp)
        print(f'{name:9} ${rec.cost_usd:.6f}  {rec.latency_s:5.2f}s  -> {resp.text[:70]}')
    except Exception as e:
        print(f'{name:9} skipped ({type(e).__name__}: {e})')
```

```python
print(tracker.table())
```

**Takeaway:** Ollama = $0 (pay in latency); cloud output tokens are usually the priciest. Edit `study_buddy/pricing.yaml` to keep rates current.
