# 02 · Chunking

**Idea:** documents are too big to embed whole, so we split them into passages. The split strategy silently caps retrieval quality. We compare three: **fixed** word windows, **recursive** (boundary-aware), and **sentence** windows.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

from rag_lab import NOTES_DIR, fixed_chunks, recursive_chunks, sentence_chunks
text = (NOTES_DIR / 'chunking_strategies.md').read_text(encoding='utf-8')
print('document words:', len(text.split()))
```

```python
f = fixed_chunks(text, chunk_size=60, overlap=10)
r = recursive_chunks(text, chunk_size=60, overlap=10)
s = sentence_chunks(text, max_sentences=3, overlap=1)
print('fixed     :', len(f), 'chunks')
print('recursive :', len(r), 'chunks')
print('sentence  :', len(s), 'chunks')
```

Look at where each strategy cuts. Fixed windows can slice mid-sentence; recursive and sentence chunking respect natural boundaries, so passages read as coherent units.

```python
print('--- fixed[0] ---\n', f[0].text[:200], '\n')
print('--- sentence[0] ---\n', s[0].text[:200])
```

**Takeaway:** chunking is a real design choice, not a detail. Boundary-aware strategies keep ideas intact; fixed windows are fast but can fragment meaning.
