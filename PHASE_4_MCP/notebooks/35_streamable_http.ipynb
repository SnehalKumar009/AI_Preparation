# 35 · Streamable HTTP transport & remote servers

stdio is perfect for local servers, but remote/shared servers need a network
transport. Modern MCP uses **Streamable HTTP** (a single HTTP endpoint that can
upgrade to server-sent events for streaming), superseding the older HTTP+SSE
pair. The **protocol messages are identical** — only the pipe changes.

> Requires `pip install mcp`. Shown as a snippet since it starts a server.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
SERVER = '''
# server.py — same tools, HTTP transport
from mcp.server.fastmcp import FastMCP
app = FastMCP('math')
@app.tool()
def add(a: int, b: int) -> int: return a + b
if __name__ == '__main__':
    app.run(transport='streamable-http')  # serves on an HTTP endpoint
'''
print(SERVER)
```

Decision guide: **stdio** for local, single-user tools (this whole lab);
**Streamable HTTP** for remote, multi-user, or independently-deployed servers.
Everything you built in notebooks 13–17 ports over unchanged.
