# 10 · Tools in depth (schemas, validation, structured results)

A tool is only as usable as its `inputSchema`. A precise JSON Schema tells the
model exactly which arguments exist, their types, and which are required —
which dramatically improves tool-calling reliability.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import MiniMCPServer
s = MiniMCPServer('files')
schema = {'type':'object','properties':{
    'path':{'type':'string','description':'path under root'},
    'max_bytes':{'type':'integer','description':'truncate length'}},
  'required':['path']}
@s.tool(name='read', description='Read a file.', input_schema=schema)
def read(path, max_bytes=1000):
    return f'would read {path} up to {max_bytes} bytes'
print(s.handle({'jsonrpc':'2.0','id':1,'method':'tools/list'})['result']['tools'][0])
```

Results are returned as **content blocks** (`{type:'text', text:...}`).
Richer servers can return multiple blocks or structured JSON as text.
