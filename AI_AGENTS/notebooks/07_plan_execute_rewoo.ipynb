# 07 · Plan-and-Execute vs ReWOO

ReAct re-plans after **every** observation — flexible but many LLM calls. **ReWOO** (*Reasoning WithOut Observation*) plans **all** tool calls up front in one shot, executes them, then composes the answer from the results. Fewer model calls, lower cost, when the plan doesn't depend on intermediate results.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
```

```python
from agents_lab import rewoo, default_tools, Tracer
from study_buddy import CostTracker

tracker = CostTracker()
tracer = Tracer()
res = rewoo('Compute 12*12 and 15*3, using the calculator for each.', tools=default_tools(), tracker=tracker, provider='ollama')
print(tracer.pretty())
print('\nANSWER:', res.answer)
print('cost:', tracker.summary())
```

The trajectory shows all tool calls were planned **before** any execution — no re-planning between them. Compare the call count to a ReAct run of the same task.

```python
from agents_lab import react

react_tracker = CostTracker()
react('Compute 12*12 and 15*3, using the calculator for each.', tools=default_tools(), tracker=react_tracker, provider='ollama')
print('ReWOO calls:', len(tracker.records))
print('ReAct calls:', len(react_tracker.records))
```

**Takeaway:** choose the planner to fit the task. ReAct adapts step-by-step; ReWOO front-loads the plan to save calls when steps are independent.
