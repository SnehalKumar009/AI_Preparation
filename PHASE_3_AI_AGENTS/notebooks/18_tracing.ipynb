# 18 · Tracing an Agent

An agent loop is invisible unless you record it. Every `agents_lab` agent writes its run into a **`Tracer`** as ordered `Step`s (thought / action / observation / answer). That makes runs debuggable: print the trajectory, count tool calls, find where it went wrong. Deep production observability (spans, dashboards, alerting) is Phase 8.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

```python
from agents_lab import Agent, Tracer, default_tools

tracer = Tracer()
agent = Agent(provider='ollama', tools=default_tools(), max_steps=6)
res = agent.run('Compute 9*9, then tell me the current time.', tracer=tracer)
print(tracer.pretty())
```

The trace is structured data, so you can analyze it — e.g. list just the tool calls, or count steps by kind.

```python
from collections import Counter

print('steps by kind:', Counter(s.kind for s in tracer))
print('tool calls   :', [s.meta.get('tool') for s in tracer.tool_calls()])
print('total steps  :', len(tracer))
```

**Takeaway:** if you can't see the trajectory, you can't debug the agent. Structured tracing is the minimum; Phase 8 turns it into full observability.
