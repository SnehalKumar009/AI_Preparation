# 06 · Temperature

**Idea:** temperature scales randomness. `0` = deterministic/greedy (same answer every time). Higher = more varied/creative (and more error-prone). Run the same creative prompt at different temperatures and watch variety change.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')
prompt = [{'role': 'user', 'content': 'Give a two-word name for a coffee shop.'}]
```

```python
for temp in [0.0, 0.7, 1.2]:
    print(f'--- temperature = {temp} ---')
    for _ in range(3):
        print('  ', llm.chat(prompt, temperature=temp).text.strip()[:40])
```

**Takeaway:** low temperature for facts/extraction/tools; higher for brainstorming. For agents you usually want low temperature = predictable.
