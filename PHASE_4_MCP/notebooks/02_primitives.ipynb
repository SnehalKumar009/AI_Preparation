# 02 · MCP primitives: tools, resources, prompts

An MCP server can expose three kinds of things:

| Primitive | Who drives it | Analogy |
|-----------|---------------|---------|
| **Tools** | model-controlled | a function the model may call |
| **Resources** | app-controlled | a file/record the host can read |
| **Prompts** | user-controlled | a reusable prompt template |

This phase focuses on **tools** (what an agent acts through); resources and
prompts are covered in notebooks 11–12.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import MiniMCPServer
s = MiniMCPServer('demo')

@s.tool(description='Say hello to NAME.')
def greet(name: str):
    return f'Hello, {name}!'

# tools/list is how a client discovers what a server offers:
print([t.to_wire()['name'] for t in s._tools.values()])
print(s.handle({'jsonrpc':'2.0','id':1,'method':'tools/list'}))
```

The server advertises each tool with a name, description, and an
`inputSchema` (JSON Schema). That schema is exactly what a model needs to
decide *when* and *how* to call the tool.
