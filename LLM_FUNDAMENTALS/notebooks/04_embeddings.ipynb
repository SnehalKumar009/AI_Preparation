# 04 · Embeddings

**Idea:** an embedding maps text to a vector so that *similar meanings sit close together*. This powers semantic search and RAG. We use local `nomic-embed-text` via Ollama.

> Requires `ollama pull nomic-embed-text` on your Ubuntu host.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

import numpy as np
from study_buddy import get_provider

ollama = get_provider('ollama')
```

```python
texts = [
    'A dog barked loudly at the mailman.',
    'The puppy made noise at the delivery person.',
    'Quarterly revenue grew by twelve percent.',
]
vecs = np.array(ollama.embed(texts))
print('embedding shape:', vecs.shape)
```

```python
def cosine(a, b):
    return float(a @ b / (np.linalg.norm(a) * np.linalg.norm(b)))

print('dog vs puppy  :', round(cosine(vecs[0], vecs[1]), 3), '(similar meaning)')
print('dog vs revenue:', round(cosine(vecs[0], vecs[2]), 3), '(unrelated)')
```

```python
# Semantic search: find the closest text to a query.
query = 'animal making sound'
q = np.array(ollama.embed([query])[0])
scores = [cosine(q, v) for v in vecs]
best = int(np.argmax(scores))
print('query   :', query)
print('best hit:', texts[best], '=>', round(scores[best], 3))
```

**Takeaway:** embeddings turn meaning into geometry. Cosine similarity ranks relevance — the engine behind RAG retrieval.
