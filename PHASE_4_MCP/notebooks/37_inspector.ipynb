# 37 · Debugging with the MCP Inspector

The **MCP Inspector** is the official interactive tool for developing servers:
it launches your server, runs the handshake, and lets you list/call tools,
read resources, and watch the raw JSON-RPC traffic — invaluable when a tool
misbehaves.

It ships via `npx` (Node), so run it from a terminal, not this cell.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
cmds = [
  '# point the Inspector at any lab server (from the PHASE_4_MCP folder):',
  'npx @modelcontextprotocol/inspector python -m mcp_lab.servers.calculator',
  'npx @modelcontextprotocol/inspector python -m mcp_lab.servers.filesystem',
  '# it opens a local web UI to list tools and call them interactively']
print('\n'.join(cmds))
```

Debugging checklist when a server is silent or broken:
1. Is anything printed to **stdout** that is not JSON? (It corrupts the stream —
   log to stderr instead; see notebook 03.)
2. Does `initialize` return your `serverInfo`?
3. Do tool `inputSchema`s match the arguments you send?
4. Are errors returned as `isError` results rather than crashes?
