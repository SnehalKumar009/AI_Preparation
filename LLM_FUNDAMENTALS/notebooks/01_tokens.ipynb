# 01 · Tokens

**Idea:** LLMs don't see characters or words — they see *tokens* (subword pieces). Everything (context window, pricing, latency) is measured in tokens.

In this notebook you will:
- Split text into tokens and inspect the token ids
- See that token count != word count
- Count tokens for a full chat message list

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

import tiktoken
from study_buddy.tokens import count_tokens, count_message_tokens
```

```python
enc = tiktoken.get_encoding('cl100k_base')
text = 'Tokenization splits text into subword pieces.'
ids = enc.encode(text)
print('token ids :', ids)
print('tokens    :', [enc.decode([i]) for i in ids])
print('words     :', len(text.split()))
print('tokens    :', len(ids))
```

```python
# Odd strings cost more tokens than you'd expect.
for s in ['hello', 'antidisestablishmentarianism', '  spaces  ', 'ChatGPT-4o', '\ud83d\ude80 emoji']:
    print(f'{count_tokens(s):>3} tokens  <- {s!r}')
```

```python
messages = [
    {'role': 'system', 'content': 'You are a concise tutor.'},
    {'role': 'user', 'content': 'Explain tokens in one sentence.'},
]
print('approx prompt tokens:', count_message_tokens(messages))
```

**Takeaway:** token count drives cost, context limits, and speed. Always think in tokens, not words.
