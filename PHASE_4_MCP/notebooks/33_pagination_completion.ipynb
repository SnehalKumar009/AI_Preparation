# 33 · Pagination & argument completion

Two UX-critical protocol details:

- **Pagination** — list operations (`tools/list`, `resources/list`) return an
  opaque `nextCursor`; the client passes it back to get the next page.
- **Completion** — `completion/complete` offers autocomplete suggestions for a
  prompt/resource argument (like tab-completion in a shell).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import make_request, make_response
# Page 1 request has no cursor; the response carries nextCursor:
print(make_request(1, 'tools/list', {}))
print(make_response(1, {'tools':[{'name':'a'}], 'nextCursor':'eyJwYWdlIjoyfQ=='}))
# Page 2 sends the cursor back:
print(make_request(2, 'tools/list', {'cursor':'eyJwYWdlIjoyfQ=='}))
```

```python
# Completion: suggest values for an argument the user is typing.
req = make_request(3, 'completion/complete', {
  'ref':{'type':'ref/prompt','name':'summarize'},
  'argument':{'name':'path','value':'not'}})
resp = make_response(3, {'completion':{'values':['notes/','notebooks/'],'hasMore':False}})
print(req); print(resp)
```

Cursors are opaque — never parse them. Completion keeps large argument spaces
(file paths, table names) usable.
