# 07 · Top-p (nucleus sampling)

**Idea:** top-p keeps the smallest set of most-likely tokens whose probabilities sum to `p`, then samples among them. `top_p=1.0` considers everything; low `top_p` restricts to the safest tokens. It's an alternative/companion knob to temperature.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')
prompt = [{'role': 'user', 'content': 'Write one imaginative sentence about the sea.'}]
```

```python
# Hold temperature fixed; sweep top_p to isolate its effect.
for tp in [0.1, 0.5, 1.0]:
    print(f'--- top_p = {tp} ---')
    for _ in range(2):
        print('  ', llm.chat(prompt, temperature=1.0, top_p=tp).text.strip()[:80])
```

**Takeaway:** typically tune *either* temperature *or* top_p, not both hard. Lower top_p = safer/more focused vocabulary.
