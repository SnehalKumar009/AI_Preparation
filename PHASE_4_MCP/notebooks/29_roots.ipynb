# 29 · Roots (client advertises filesystem scope)

**Roots** are how a *client* tells servers which directories/URIs they may
operate on. The server requests `roots/list`; the client returns its roots.
This is MCP's own sandboxing handshake — complementary to the server-side
`resolve_within_root` guard from notebook 24 (defense in depth: the client
declares scope, the server still enforces it).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab.advanced import Session

session = Session(roots=[
    {'uri':'file:///Users/me/project','name':'project'},
    {'uri':'file:///Users/me/notes','name':'notes'}])

@session.tool(description='Report the roots the client exposes.')
def scope(ctx):
    roots = ctx.list_roots()
    return 'allowed roots:\n' + '\n'.join(r['uri'] for r in roots)

print(session.call_tool('scope'))
```

A well-behaved server intersects its own sandbox with the client's roots and
refuses anything outside both. Clients may also send `roots/list_changed`.
