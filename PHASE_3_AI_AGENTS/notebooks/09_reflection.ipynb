# 09 · Reflection / Self-Critique

A first draft is rarely the best. **Reflection** adds a critic step: the model reviews its own answer, names concrete problems, then revises. `reflect_and_revise` loops draft → critique → revise until the critic says `OK` or a round cap is hit.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

A single critique pass over a deliberately weak answer:

```python
from agents_lab import reflect

task = 'Explain why the sky is blue.'
weak = 'Because it reflects the ocean.'
print(reflect(task, weak, provider='ollama'))
```

Now the full loop: draft, critique, revise. The trajectory shows each stage.

```python
from agents_lab import reflect_and_revise, Tracer

res = reflect_and_revise('Write a one-paragraph explanation of what a vector embedding is.', provider='ollama', max_rounds=2)
for step in res.steps:
    print(f"[{step.meta.get('stage', step.kind)}] {step.content[:160]}")
print('\nFINAL:\n', res.answer)
```

**Takeaway:** reflection is a cheap, powerful quality boost — let the model grade and fix its own work before returning.
