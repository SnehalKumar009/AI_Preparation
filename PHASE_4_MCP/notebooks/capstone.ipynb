# Capstone · A provider-interchangeable MCP agent with a cost meter

Combine everything: an agent over **all five servers** (plus code-search),
runnable on any provider, with a live **USD cost meter** from
`study_buddy.CostTracker`. Ask it Project 6's question and watch it choose
tools across servers.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from study_buddy import get_provider
PROVIDER = 'ollama'  # swap: openai | deepseek | anthropic | oxalpha | ollama
llm = get_provider(PROVIDER)
print('using provider:', PROVIDER)
```

```python
from study_buddy import CostTracker
from mcp_lab.agent_bridge import build_agent

tracker = CostTracker()
with build_agent(
        ['filesystem','git','sqlite','terminal','calculator','code_search'],
        provider=PROVIDER, model=None, max_steps=8, tracker=tracker) as agent:
    res = agent.run('Find every TODO in the repository, then tell me how many users are in the database.')
    print(res.answer)
    print('--- trajectory ---')
    for s in res.steps:
        print(f'{s.kind:12} {getattr(s, "name", "")}')

print(f'\nTotal cost: ${tracker.total_usd:.6f}  ({PROVIDER})')
```
