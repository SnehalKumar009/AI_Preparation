# 12 · Prompts in depth

**Prompts** are *user-controlled*, reusable, parameterized templates a server
offers (e.g. a `/summarize` slash-command). The host lists them with
`prompts/list` and fetches a filled one with `prompts/get`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
example = {'prompts':[{'name':'summarize','description':'Summarize a file',
  'arguments':[{'name':'path','required':True}]}]}
filled = {'messages':[{'role':'user','content':{'type':'text',
  'text':'Summarize the file at notes/rag_overview.md in 3 bullets.'}}]}
import json; print(json.dumps(example, indent=2)); print(json.dumps(filled, indent=2))
```

Prompts keep expert phrasings on the server side so every host gets the same
high-quality template without copy-pasting.
