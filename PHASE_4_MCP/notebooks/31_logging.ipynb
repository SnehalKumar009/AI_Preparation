# 31 · Server logging (`notifications/message`)

Servers report diagnostics to the host with `notifications/message` at a log
**level** (debug/info/warning/error). This is out-of-band from tool *results*
— it feeds the host's log UI, not the model's context.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab.advanced import Session
session = Session()

@session.tool(description='Do work while logging.')
def work(ctx, x):
    ctx.log('info', f'received x={x}')
    if x < 0:
        ctx.log('warning', 'negative input, using abs()')
        x = abs(x)
    return x * 2

print('result:', session.call_tool('work', {'x':-5}))
for n in session.notifications:
    if n['method']=='notifications/message':
        print(n['params'])
```
