# 04 · ReAct — Reason + Act

Native tool calling hides the model's reasoning. **ReAct** makes it explicit with a text protocol:

```
Thought: ...
Action: <tool>
Action Input: {json}
Observation: <result>   # filled in by us
... repeat ...
Final Answer: ...
```

Because it's just text, it works on **any** provider — the same portable idea as `rag_lab.agentic`. Seeing the Thought/Action/Observation trace is the whole point.

> Runs on Ollama.

````python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
````

````python
from agents_lab import react, default_tools, Tracer

tracer = Tracer()
res = react('What is 15 times 12, then add 8?', tools=default_tools(), provider='ollama', tracer=tracer)
print(tracer.pretty())
print('\nANSWER:', res.answer)
````

Compare with the native-tool `Agent` from notebook 01: same result, but ReAct exposes the reasoning between actions. That transparency is useful for debugging and for models without native tool calling.

````python
print('tool actions taken:')
for step in tracer.tool_calls():
    print(' ', step.content)
````

**Takeaway:** ReAct interleaves *reasoning* and *acting* in plain text. It's the most widely used agent pattern and the mental model behind most frameworks.
