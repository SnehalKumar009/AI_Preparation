# 09 · Prompt Engineering

**Idea:** how you *ask* changes the answer. Compare **zero-shot**, **few-shot** (examples), and a **structured/role** prompt on the same classification task.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')
review = 'The battery dies in an hour but the screen is gorgeous.'
```

```python
# 1) Zero-shot
zero = [{'role': 'user', 'content': f'Sentiment (positive/negative/mixed): {review}'}]
print('zero-shot :', llm.chat(zero, temperature=0).text.strip())
```

```python
# 2) Few-shot: teach the format by example
few = [
    {'role': 'user', 'content': 'Text: Great value. Label: positive'},
    {'role': 'user', 'content': 'Text: Broke on day one. Label: negative'},
    {'role': 'user', 'content': f'Text: {review} Label:'},
]
print('few-shot  :', llm.chat(few, temperature=0).text.strip())
```

```python
# 3) Structured system prompt: constrain output tightly
structured = [
    {'role': 'system', 'content': 'You are a classifier. Reply with exactly one word: positive, negative, or mixed.'},
    {'role': 'user', 'content': review},
]
print('structured:', llm.chat(structured, temperature=0).text.strip())
```

**Takeaway:** examples + explicit constraints beat vague asks. This sets up structured output (topic 11).
