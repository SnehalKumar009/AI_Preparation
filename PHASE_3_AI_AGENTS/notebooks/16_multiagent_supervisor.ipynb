# 16 · Multi-Agent — Supervisor + Workers

One agent is often not enough. A **Supervisor** routes each task to the most suitable **Worker** (a named specialist) and returns its answer — the delegation / handoff pattern. Deep orchestration (state machines, checkpoints, durable workflows) is Phase 5; this is the light version.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

Define two specialist workers — a math agent (with the calculator) and a writing agent — then let a supervisor route tasks to them.

```python
from agents_lab import Agent, Worker, Supervisor, default_tools

math_agent = Worker('math', 'Solves arithmetic and numeric problems using a calculator.',
                    Agent(provider='ollama', tools=default_tools()))
writer_agent = Worker('writer', 'Writes and edits short prose, summaries, and explanations.',
                      Agent(provider='ollama', tools=default_tools(), system='You are a concise writer.'))

boss = Supervisor([math_agent, writer_agent], provider='ollama')
```

```python
for task in ['What is 144 divided by 12?', 'Write a one-line tagline for a coffee shop.']:
    name, answer = boss.run(task)
    print(f'[routed to: {name}] {answer}\n')
print('routing cost:', boss.tracker.summary())
```

**Takeaway:** a supervisor turns a pool of specialists into one system — route by capability, delegate, collect. The foundation for multi-agent orchestration in Phase 5.
