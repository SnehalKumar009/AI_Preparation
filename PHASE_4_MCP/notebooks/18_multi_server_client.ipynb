# 18 · Multi-server client (aggregation & namespacing)

A real host connects to *several* servers at once and presents their tools as
one menu. `MCPClient` spawns one client per server and namespaces each tool
as `server__tool` (the `__` separator keeps names valid for provider
tool-calling, which forbids dots).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import MCPClient
from mcp_lab.client import lab_servers
with MCPClient(lab_servers(['calculator','filesystem','sqlite'])) as mcp:
    for t in mcp.list_tools():
        print(t['name'])
    print('->', mcp.call_tool('calculator__evaluate', {'expression':'6*7'}))
    print('->', mcp.call_tool('sqlite__list_tables', {}))
```

Namespacing prevents collisions (two servers could both have a `search`
tool) and lets the client route a call to the right subprocess.
