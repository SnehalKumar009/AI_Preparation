# 10 · Reflexion — Reflection with Memory

Plain reflection fixes one answer. **Reflexion** adds an *episodic memory of lessons*: after each failed attempt an **evaluator** gives feedback, the agent writes a one-line self-reflection, and that lesson is carried into the next attempt so it stops repeating the mistake.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

We need an `evaluator(answer) -> (passed, feedback)`. Here a simple checker requires the answer to contain a specific keyword, simulating a task constraint the model keeps missing.

```python
from agents_lab import Reflexion

def evaluator(answer):
    if 'haiku' in answer.lower() and answer.count('\n') >= 2:
        return True, 'looks like a 3-line haiku'
    return False, 'must be a 3-line haiku and mention the word "haiku"'

engine = Reflexion(provider='ollama', max_trials=3)
res = engine.run('Write a short poem about autumn.', evaluator)
print('lessons learned:', engine.reflections)
print('\nFINAL:\n', res.answer)
```

Inspect the trial-by-trial trajectory: attempt → eval → reflection → next attempt.

```python
for step in res.steps:
    print(f"[trial {step.meta.get('trial','?')}] {step.kind}: {step.content[:120]}")
```

**Takeaway:** Reflexion = reflection + memory across attempts. The agent learns *within* a task by carrying lessons forward — a bridge to the full memory taxonomy in Phase 6.
