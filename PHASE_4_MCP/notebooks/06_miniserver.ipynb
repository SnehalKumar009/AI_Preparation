# 06 · A minimal MCP server, from scratch

`MiniMCPServer` is the entire server side in one small class: register tools,
answer `initialize` / `tools/list` / `tools/call`, and serve over stdio. The
five real servers are just this class with domain tools attached.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import MiniMCPServer
srv = MiniMCPServer('math')

@srv.tool(description='Add two numbers.', input_schema={'type':'object','properties':{'a':{'type':'number'},'b':{'type':'number'}},'required':['a','b']})
def add(a, b):
    return a + b

# Drive it directly (no subprocess) to see request -> response:
print(srv.handle({'jsonrpc':'2.0','id':1,'method':'tools/call','params':{'name':'add','arguments':{'a':2,'b':3}}}))
```

A tool that raises does **not** crash the loop: the error comes back as an
`isError` result so the model can read it as an observation and recover.
Run a server as a process with `python -m mcp_lab.servers.calculator`.
