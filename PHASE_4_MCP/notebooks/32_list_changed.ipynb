# 32 · List-changed & resource subscriptions

Servers are not static. When a server's tools or resources change, it emits
`notifications/tools/list_changed` (or `resources/list_changed`) so the host
re-fetches. For individual resources, a client can `resources/subscribe` and
receive `notifications/resources/updated` when that resource changes.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab.advanced import Session
session = Session()
print('before:', [t['name'] for t in session.list_tools()])

@session.tool(description='A newly added tool.')
def added(ctx):
    return 'hi'
session.emit_list_changed('tools')  # tell the client to refresh

print('after :', [t['name'] for t in session.list_tools()])
print('notifs:', [n['method'] for n in session.notifications])
```

```python
# Subscription message shapes (what a client would send / receive):
from mcp_lab import make_request, make_notification
print(make_request(1, 'resources/subscribe', {'uri':'file:///data.csv'}))
print(make_notification('notifications/resources/updated', {'uri':'file:///data.csv'}))
```

Subscriptions let a host keep a live view of a resource without polling.
