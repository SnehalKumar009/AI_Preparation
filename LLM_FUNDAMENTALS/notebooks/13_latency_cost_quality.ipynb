# 13 · Latency vs Cost vs Quality

**Idea:** model selection is a trade-off. Local is free but slower; small cloud models are cheap and fast; reasoning models are pricey but stronger. Benchmark them on one task and read the table.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider, CostTracker

task = [{'role': 'user', 'content': 'Summarize what an embedding is in exactly one sentence.'}]
tracker = CostTracker()
candidates = [
    ('ollama', None),
    ('openai', 'gpt-4o-mini'),
    ('gemini', 'gemini-2.0-flash'),
    ('deepseek', 'deepseek-reasoner'),
    ('anthropic', 'claude-haiku-4-5-20251001'),
]
```

```python
answers = {}
for name, model in candidates:
    try:
        resp = get_provider(name).chat(task, model=model, temperature=0)
        tracker.add(resp)
        answers[name] = resp.text.strip()
    except Exception as e:
        print(f'{name} skipped: {e}')
```

```python
print(tracker.table())
print()
for name, text in answers.items():
    print(f'[{name}] {text}\n')
```

**Takeaway:** there is no single 'best' model — pick per task. Cheap+fast for high volume, strong+slow for hard problems. Route intelligently.
