# 15 · SQLite server (schema + safe SELECT)

Read-only DB access: the connection is opened `mode=ro` and `run_select`
rejects anything that is not a single `SELECT` (notebook 25 covers the idea).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.sqlite']
```

```python
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print('tables', c.call_tool('list_tables', {}))
    print('users', c.call_tool('run_select', {'query': 'SELECT * FROM users'}))
    print('blocked', c.call_tool('run_select', {'query': 'DELETE FROM users'}))
```
