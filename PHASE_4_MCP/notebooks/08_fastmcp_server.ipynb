# 08 · A server with the official SDK (FastMCP)

Now that the mechanics are clear, the official `mcp` SDK removes the
boilerplate. `FastMCP` uses decorators much like our `MiniMCPServer`.

> Requires `pip install mcp`. The cell degrades gracefully if it is absent.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
try:
    from mcp.server.fastmcp import FastMCP
    app = FastMCP('math')
    @app.tool()
    def add(a: int, b: int) -> int:
        'Add two integers.'
        return a + b
    print('FastMCP server built with tool: add')
except ImportError:
    print('Install the official SDK to run this:  pip install mcp')
```

Compare with notebook 06: same idea (register a typed tool), but the SDK
derives the JSON schema from the type hints and handles the transport.
