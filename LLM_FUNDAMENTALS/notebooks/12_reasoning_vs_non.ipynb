# 12 · Reasoning vs Non-reasoning Models

**Idea:** reasoning models (e.g. `deepseek-reasoner`) spend extra tokens 'thinking' before answering, which helps on multi-step logic/math — at higher cost and latency. Compare against a standard model (local `qwen3-coder:30b`) on a tricky puzzle.

> Needs `DEEPSEEK_API_KEY` in `.env`.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider, CostTracker
tracker = CostTracker()

puzzle = ('A bat and ball cost $1.10 total. The bat costs $1.00 more than the ball. '
          'How much is the ball? Think step by step, then give the final number.')
messages = [{'role': 'user', 'content': puzzle}]
```

```python
# Non-reasoning: local qwen3-coder
r1 = get_provider('ollama').chat(messages, temperature=0)
tracker.add(r1)
print('qwen3-coder (non-reasoning):\n', r1.text[:400])
```

```python
# Reasoning: deepseek-reasoner
try:
    r2 = get_provider('deepseek').chat(messages, model='deepseek-reasoner')
    tracker.add(r2)
    print('deepseek-reasoner:\n', r2.text[:400])
except Exception as e:
    print('deepseek skipped:', e)
```

```python
print(tracker.table())
print('\nNote how the reasoning model uses more output tokens (its thinking).')
```

**Takeaway:** use reasoning models for hard logic/planning; use cheaper/faster models for routine tasks. The right answer is $0.05.
