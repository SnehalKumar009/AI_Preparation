# 21 · Demo: "Find every TODO in the repository"

The canonical MCP demo: the agent is given filesystem + terminal tools and
**decides** how to find TODOs — no hardcoded pipeline. This is Project 6's
headline query.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from study_buddy import get_provider
PROVIDER = 'ollama'  # swap: openai | deepseek | anthropic | oxalpha | ollama
llm = get_provider(PROVIDER)
print('using provider:', PROVIDER)
```

```python
from mcp_lab.agent_bridge import build_agent
with build_agent(['filesystem','terminal','code_search'], provider=PROVIDER, max_steps=6) as agent:
    res = agent.run('Find every TODO or FIXME comment in this repository and list the files.')
    print(res.answer)
    print('steps:', len(res.steps), 'tool calls:', res.n_tool_calls)
```
