# 05 · The `initialize` handshake & capability negotiation

Every MCP session starts with `initialize`: the client sends its protocol
version + capabilities, the server replies with **its** version, capabilities,
and `serverInfo`. The client then sends the `notifications/initialized`
notification and normal traffic begins.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import MiniMCPServer, PROTOCOL_VERSION
s = MiniMCPServer('demo')
req = {'jsonrpc':'2.0','id':1,'method':'initialize','params':{'protocolVersion':PROTOCOL_VERSION,'capabilities':{}}}
print(s.handle(req)['result'])
```

The reply's `capabilities` announces which primitives the server supports
(`tools`, `resources`, `prompts`). A host uses this to decide what to show.
