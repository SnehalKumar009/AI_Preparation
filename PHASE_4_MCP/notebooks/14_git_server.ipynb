Read-only git introspection via GitPython over `MCP_GIT_ROOT`. Requires
`pip install GitPython` and a git repo. Never commits, checks out, or pushes.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.git']
```

```python
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print('log', c.call_tool('git_log', {'max_count': 3}))
```
