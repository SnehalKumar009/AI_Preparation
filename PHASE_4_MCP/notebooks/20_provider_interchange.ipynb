# 20 · Provider interchangeability across MCP

Because the agent uses the provider's **normalized** `tool_calls`, swapping
providers is a one-string change. The MCP layer is entirely provider-agnostic.

| Provider | Tool calling | Notes |
|----------|--------------|-------|
| ollama (`qwen3-coder`) | yes | local default |
| openai | yes | `gpt-4o-mini` |
| deepseek | yes | OpenAI-compatible |
| anthropic | yes | different wire format, handled in provider |
| oxalpha | depends | routed model must support tools |

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab.agent_bridge import build_agent
TASK = 'Use the calculator to compute (18+24)/6.'
for provider in ['ollama']:  # add 'openai','deepseek',... when keys are set
    try:
        with build_agent(['calculator'], provider=provider, max_steps=3) as agent:
            print(provider, '->', agent.run(TASK).answer)
    except Exception as e:
        print(provider, 'skipped:', e)
```

Add more providers to the loop once their keys are in
`PHASE_1_LLM_FUNDAMENTALS/.env`. Missing keys skip cleanly.
