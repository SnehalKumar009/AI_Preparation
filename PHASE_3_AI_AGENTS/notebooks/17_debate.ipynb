# 17 · Multi-Agent — Handoffs & Debate

Two agents can be better than one. In a **debate**, opposing personas argue for a few rounds and a **judge** picks the best-supported answer — which reduces one-sided mistakes on open-ended or contested questions.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

```python
from agents_lab import debate
from study_buddy import CostTracker

tracker = CostTracker()
verdict = debate(
    'Should a small startup build its own vector database or use a managed one?',
    provider='ollama', rounds=2, personas=('Build', 'Buy'), tracker=tracker,
)
print('VERDICT:\n', verdict)
print('\ncost:', tracker.summary())
```

**Takeaway:** structured disagreement (debate) plus a judge is a simple way to raise answer quality on contested questions — a multi-agent critique pattern you'll scale up in Phase 5.
