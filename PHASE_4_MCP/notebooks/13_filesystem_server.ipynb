through `resolve_within_root`, so `../` traversal is blocked (notebook 24).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.filesystem']
```

```python
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print('list .', c.call_tool('list_dir', {'path': '.'}))
    print('*.py', c.call_tool('search_files', {'pattern': '*.py', 'path': 'mcp_lab'}))
```
