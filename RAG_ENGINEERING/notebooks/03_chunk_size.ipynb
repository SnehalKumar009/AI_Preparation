# 03 · Chunk Size

**Idea:** small chunks retrieve *precisely* but may lack context; large chunks keep context but *dilute* the signal. We measure the trade-off by retrieving the same query at several chunk sizes. Embeddings via **Ollama**.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

from rag_lab import NOTES_DIR, fixed_chunks, NumpyStore
text = (NOTES_DIR / 'vector_search.md').read_text(encoding='utf-8')
query = 'How does approximate nearest neighbor search speed things up?'
```

```python
for size in [30, 80, 160, 320]:
    store = NumpyStore()
    store.add(fixed_chunks(text, chunk_size=size, overlap=size // 5))
    top = store.search(query, k=1)[0]
    print(f'size={size:>3} | chunks={len(store):>2} | top score={top.score:.3f} | '
          f'top words={len(top.chunk.text.split())}')
```

Very small chunks often score high but return too little to answer; very large chunks bury the relevant sentence. The sweet spot is usually a moderate size tuned on real queries — not guessed.

```python
store = NumpyStore(); store.add(fixed_chunks(text, chunk_size=80, overlap=16))
print(store.search(query, k=1)[0].chunk.text)
```

**Takeaway:** chunk size trades precision against context. Start near 200 tokens and tune against an evaluation set (notebook 21).
