# 36 · Auth for remote MCP servers (OAuth 2.1)

Local stdio servers inherit the user's own permissions — no auth needed. But a
**remote** server exposed over HTTP must authorize callers. MCP specifies
**OAuth 2.1**: the server advertises auth metadata, the client obtains a token,
and every request carries `Authorization: Bearer <token>`.

This is a protocol/deployment concern, not something the stdio lab runs — the
flow and headers are shown below.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
flow = [
  '1. client GET /.well-known/oauth-authorization-server  -> discovery',
  '2. user authorizes; client receives an authorization code',
  '3. client exchanges code at /token -> access_token (+ refresh)',
  '4. every MCP request: header Authorization: Bearer <access_token>',
  '5. server validates scope/audience before running any tool']
print('\n'.join(flow))
example_headers = {'Authorization':'Bearer eyJ...','MCP-Protocol-Version':mcp_lab.PROTOCOL_VERSION}
print(example_headers)
```

Rules of thumb: never put tokens in tool arguments or logs (notebook 26);
scope tokens to least privilege; prefer short-lived tokens with refresh.
