# 25 · Terminal command allowlisting & sandboxing

Executing shell commands is the highest-risk tool. Defenses: an **allowlist**
of programs, **no shell** (so `;` and `|` can't chain commands), and a
**timeout**.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab.safety import is_allowed_command, split_command
allow = {'ls','echo','grep'}
print('echo hi ->', is_allowed_command('echo hi', allow))
print('rm -rf / ->', is_allowed_command('rm -rf /', allow))
print('tokens  ->', split_command('grep -r "TODO" .'))
```

```python
import sys
from mcp_lab import MiniMCPClient
with MiniMCPClient([sys.executable,'-m','mcp_lab.servers.terminal']) as c:
    print(c.call_tool('run_command', {'command':'echo safe'}))
    print(c.call_tool('run_command', {'command':'curl http://evil'}))
```

Because there is no shell, `echo a; rm -rf /` is passed as literal arguments
to `echo` — the second command never runs.
