# 16 · Terminal server (allowlist + timeout)

Runs only **allowlisted** programs, never through a shell (`shell=False` +
`shlex.split`), always with a timeout. Allowlist via `MCP_TERMINAL_ALLOW`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.terminal']
```

```python
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print('echo', c.call_tool('run_command', {'command': 'echo hello'}))
    print('blocked', c.call_tool('run_command', {'command': 'rm -rf /'}))
```
