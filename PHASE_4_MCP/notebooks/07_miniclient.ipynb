# 07 · A minimal MCP client, from scratch

`MiniMCPClient` launches a server subprocess and speaks JSON-RPC to it:
`initialize` → `tools/list` → `tools/call`. This is what every MCP host does
under the hood.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
with MiniMCPClient([sys.executable, '-m', 'mcp_lab.servers.calculator']) as c:
    print('serverInfo:', c.server_info.get('serverInfo'))
    print('tools     :', [t['name'] for t in c.list_tools()])
    print('2*(3+4)   :', c.call_tool('evaluate', {'expression':'2*(3+4)'}))
```

The context manager guarantees the subprocess is shut down. Notifications
from the server are skipped while waiting for the response matching our `id`.
