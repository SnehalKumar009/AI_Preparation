# 07 · Similarity Search

**Idea:** put it together — embed a corpus, embed a query, return the **top-k** nearest chunks. This is dense retrieval, the default first stage of RAG. `NumpyStore` does it with one normalized matrix multiply.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

from rag_lab import load_markdown, NumpyStore
store = NumpyStore()
store.add(load_markdown())
print('indexed chunks:', len(store))
```

```python
query = 'how do I combine keyword and vector search?'
for h in store.search(query, k=4):
    print(f'{h.score:.3f}  {h.chunk.source} :: {h.chunk.meta["heading"]}')
```

`context_for` formats the top hits for prompt injection, tagging each with its source so the generator can cite it (notebook 17).

```python
print(store.context_for(query, k=2))
```

**Takeaway:** dense top-k retrieval is the RAG baseline. It captures meaning but can miss exact tokens — the gap we close with keyword and hybrid search next.
