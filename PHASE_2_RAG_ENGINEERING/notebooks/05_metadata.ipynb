# 05 · Metadata

**Idea:** chunks are more than text. Attaching **source**, **heading**, and **tags** lets you *filter before ranking* — restrict search to a document or topic, then sort by similarity. Our corpus loader parses front-matter into metadata automatically.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

from rag_lab import load_markdown, NumpyStore
chunks = load_markdown()
print('sections:', len(chunks))
c = chunks[0]
print('source :', c.source)
print('heading:', c.meta['heading'])
print('tags   :', c.meta.get('tags'))
```

```python
store = NumpyStore()
store.add(chunks)
query = 'how big should a passage be?'
print('--- unfiltered ---')
for h in store.search(query, k=3):
    print(f'{h.score:.3f}  {h.chunk.source} :: {h.chunk.meta["heading"]}')
```

Now the same query, restricted to chunks tagged `chunk-size` via a metadata predicate. Filtering narrows the candidate set *before* similarity ranking.

```python
def has_tag(chunk, tag):
    return tag in (chunk.meta.get('tags') or [])

print('--- filtered: tag=chunk-size ---')
for h in store.search(query, k=3, where=lambda c: has_tag(c, 'chunk-size')):
    print(f'{h.score:.3f}  {h.chunk.source} :: {h.chunk.meta["heading"]}')
```

**Takeaway:** metadata turns a flat index into a filterable one. Pre-filtering by source/tag/date improves precision and enables access control (notebook 26).
