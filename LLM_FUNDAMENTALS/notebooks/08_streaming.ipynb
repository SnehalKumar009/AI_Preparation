# 08 · Streaming

**Idea:** instead of waiting for the whole answer, the model emits tokens as they're generated. Streaming makes apps feel fast and is essential UX for chat/agents. Every provider in `study_buddy` exposes `.stream()`.

```python
import sys, pathlib, time
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')
```

```python
prompt = [{'role': 'user', 'content': 'List 5 uses for embeddings, one per line.'}]
start = time.perf_counter()
first_token_at = None
for chunk in llm.stream(prompt, temperature=0):
    if first_token_at is None:
        first_token_at = time.perf_counter() - start
    print(chunk, end='', flush=True)
print(f'\n\n[time to first token: {first_token_at:.2f}s]')
```

**Takeaway:** streaming improves perceived latency (time-to-first-token matters more than total time for UX). The Streamlit capstone streams live.
