# 05 · Roles (system / user / assistant)

**Idea:** a chat request is a list of messages with roles. The **system** message sets behavior, **user** is the human, **assistant** is the model's prior replies (memory). Changing the system prompt changes everything.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'study_buddy').exists())
sys.path.insert(0, str(root))

from study_buddy import get_provider
llm = get_provider('ollama')  # swap to 'openai' | 'gemini' | 'deepseek'
```

```python
question = 'How do I center a div?'
for persona in ['You are a terse senior engineer. One line only.',
                'You are a friendly teacher who explains with an analogy.']:
    msgs = [{'role': 'system', 'content': persona},
            {'role': 'user', 'content': question}]
    print('SYSTEM:', persona)
    print('REPLY :', llm.chat(msgs, temperature=0).text[:200])
    print('-' * 60)
```

```python
# assistant turns act as memory: the model uses prior context.
conversation = [
    {'role': 'system', 'content': 'You are a helpful assistant.'},
    {'role': 'user', 'content': 'My name is Sanket.'},
    {'role': 'assistant', 'content': 'Nice to meet you, Sanket!'},
    {'role': 'user', 'content': 'What is my name?'},
]
print(llm.chat(conversation, temperature=0).text)
```

**Takeaway:** the system message is your steering wheel; assistant turns are the conversation's memory. Prompt engineering (topic 09) builds on this.
