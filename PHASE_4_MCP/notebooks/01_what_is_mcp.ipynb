# 01 · What is MCP (host / client / server)

The **Model Context Protocol** standardizes how an LLM application discovers
and calls capabilities that live *outside* the model. Three roles:

- **Host** — the app the user talks to (an editor, a chat UI, an agent).
- **Client** — the connector inside the host that speaks MCP to one server.
- **Server** — an independent process exposing *tools*, *resources*, *prompts*.

One host runs many clients, each connected to one server. This notebook just
establishes the mental model; later notebooks build every piece from scratch.
