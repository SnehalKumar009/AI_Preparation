# 04 · JSON-RPC 2.0 basics

MCP is JSON-RPC 2.0. Three message shapes:

- **request** — has an `id`, expects a matching response.
- **notification** — no `id`, fire-and-forget (e.g. `notifications/initialized`).
- **response** — carries `result` **or** `error` for a given `id`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from mcp_lab import make_request, make_notification, make_response, make_error
print(make_request(1, 'tools/list'))
print(make_notification('notifications/initialized'))
print(make_response(1, {'tools': []}))
print(make_error(1, -32601, 'Method not found'))
```

Standard error codes: `-32700` parse, `-32600` invalid request, `-32601`
method not found, `-32602` invalid params, `-32603` internal. Servers reuse
these so any compliant client understands failures.
