# 21 · Multi-Agent — Group Chat (round-robin)

A **Pipeline** is one pass in a fixed order. A **GroupChat** instead loops:
members take turns over a *shared transcript* for several rounds, so they can
build on and correct each other before a final summary distills the result.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
import agents_lab  # noqa: F401  (bootstraps sibling study_buddy + rag_lab)
```

Three personas discuss a design question for two rounds; a summarizer returns
the final answer.

```python
from agents_lab import Agent, Worker, GroupChat

members = [
    Worker('Optimist', 'argues for the boldest option', Agent(provider='ollama', system='You are an optimistic architect.')),
    Worker('Skeptic',  'stress-tests risks and edge cases', Agent(provider='ollama', system='You are a cautious skeptic.')),
    Worker('Pragmatist','favors the simplest shippable choice', Agent(provider='ollama', system='You are a pragmatic engineer.')),
]
chat = GroupChat(members, rounds=2, provider='ollama')
```

```python
final = chat.run('Should a small startup build on microservices or a monolith?')
print('FINAL:\n', final)
```

```python
# Inspect the raw discussion (no summary) to see the round-robin turns:
transcript = chat.run('Same question, show the debate.', summarize=False)
print(transcript[:800])
```

**Takeaway:** group chat suits open-ended problems where perspectives should
collide before converging. Pipelines suit well-defined, ordered work; group
chat suits deliberation.
