# 05 · Chain-of-Thought & Self-Consistency

Before an agent can *plan*, it needs to *reason*. **Chain-of-Thought (CoT)** asks the model to work step by step before answering. **Self-Consistency** samples several independent reasoning paths and takes a **majority vote**, trading cost for accuracy on tricky problems.

> Runs on Ollama.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
```

```python
from agents_lab import chain_of_thought
from study_buddy import CostTracker

q = 'A shop had 15 apples, sold 9, then received 2 boxes of 6. How many apples now?'
tracker = CostTracker()
answer, reasoning = chain_of_thought(q, provider='ollama', tracker=tracker)
print('REASONING:\n', reasoning)
print('\nANSWER:', answer)
```

**Self-consistency:** sample several paths at temperature > 0 and vote. Watch the individual answers — the majority is usually right even when a single path errs.

```python
from agents_lab import self_consistency

voted, samples = self_consistency(q, n=5, provider='ollama', temperature=0.8)
print('samples:', samples)
print('voted  :', voted)
```

**Takeaway:** CoT surfaces reasoning; self-consistency makes it more reliable by voting across paths. These primitives feed directly into planning and tree search.
