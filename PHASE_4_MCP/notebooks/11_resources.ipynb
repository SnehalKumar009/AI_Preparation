# 11 · Resources in depth

**Resources** are *app-controlled* data the host can read and (optionally) put
into context — files, DB rows, API payloads — addressed by URI. Unlike tools,
the model does not decide to call them; the host chooses what to attach.

Our minimal server returns an empty `resources/list`; here is the shape a
fuller server would expose.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
example = {'resources': [
  {'uri':'file:///notes/rag_overview.md','name':'RAG overview','mimeType':'text/markdown'},
  {'uri':'sqlite:///users','name':'users table','mimeType':'application/json'}]}
import json; print(json.dumps(example, indent=2))
```

A host reads a resource with `resources/read` and may pin it into the prompt.
Think of tools as *verbs* and resources as *nouns*.
