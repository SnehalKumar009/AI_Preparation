# 09 · An MCP client with the SDK

The SDK also ships a client that spawns a server over stdio and discovers its
tools — the same handshake you built by hand in notebook 07.

> Requires `pip install mcp`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
# Sketch of the official async client API (runs when `mcp` is installed):
SNIPPET = '''
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client
params = StdioServerParameters(command='python', args=['-m','mcp_lab.servers.calculator'])
async with stdio_client(params) as (r, w):
    async with ClientSession(r, w) as session:
        await session.initialize()
        tools = await session.list_tools()
        print([t.name for t in tools.tools])
'''
print(SNIPPET)
```

The shape is identical to `MiniMCPClient`; only the async plumbing differs.
For the rest of the lab we keep using `MiniMCPClient` so cells stay sync.
