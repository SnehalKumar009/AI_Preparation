# 15 · Agentic RAG — Retrieval as a Tool

Phase 2 built retrievers; here retrieval becomes just another **tool** an agent chooses to call. `make_rag_tool` wraps any `rag_lab` retriever into a `Tool`; `build_rag_agent` returns an `Agent` that decides *whether, what, and how many times* to search, then answers grounded in what it retrieved.

> Runs on Ollama (uses local embeddings). Reuses the Phase 2 corpus.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'agents_lab').exists())
sys.path.insert(0, str(root))
```

Build a retriever over the Phase 2 multi-hop corpus (reused directly from `RAG_ENGINEERING/multihop`).

```python
from rag_lab import load_markdown, NumpyStore

corpus_dir = root.parent / 'RAG_ENGINEERING' / 'multihop'
store = NumpyStore(provider='ollama')
store.add(load_markdown(directory=corpus_dir))
print('chunks indexed:', len(store))
```

Give the agent a question that needs retrieval. Watch it call the `search` tool, then answer with `[source]` citations.

```python
from agents_lab import build_rag_agent, Tracer

rag_agent = build_rag_agent(store, provider='ollama', k=4, max_steps=5)
tracer = Tracer()
res = rag_agent.run('What companies were founded by Sam\'s college roommate?', tracer=tracer)
print(tracer.pretty())
print('\nANSWER:\n', res.answer)
print('\ncost:', res.tracker.summary())
```

A question needing **no** retrieval — a good agent answers directly instead of searching.

```python
res2 = rag_agent.run('What is 7 times 6?')
print('tool calls:', res2.n_tool_calls)
print('ANSWER:', res2.answer)
```

**Takeaway:** agentic RAG hands *control* of retrieval to the model — retrieval is a tool, not a fixed pipeline stage. This is the backbone of the capstone.
