# 22 · Code-search server (the Code RAG fold-in)

This closes **Project 3**: index a codebase and answer *"where is X?"*. It
reuses Phase 2 `rag_lab` for chunking + semantic retrieval, with a substring
`grep_code` fallback when embeddings (Ollama `nomic-embed-text`) are absent.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import sys
from mcp_lab import MiniMCPClient
CMD = [sys.executable, '-m', 'mcp_lab.servers.code_search']
with MiniMCPClient(CMD) as c:
    print('tools:', [t['name'] for t in c.list_tools()])
    print(c.call_tool('grep_code', {'text':'def build_server'}).splitlines()[0])
    # semantic search (needs Ollama embeddings; falls back to grep):
    print(c.call_tool('search_code', {'query':'where is path traversal blocked', 'k':2})[:200])
```

Point it at any repo with `MCP_CODE_ROOT=/path python -m mcp_lab.servers.code_search`.
Semantic search mirrors an AI coding assistant's retrieval step.
