# 06 · Cosine Similarity

**Idea:** cosine similarity measures the *angle* between vectors, ignoring length. We derive it by hand and show that once vectors are L2-normalized, cosine is just the **dot product** — which is why retrieval is one matrix multiply.

```python
import sys, pathlib
p = pathlib.Path.cwd()
root = next(x for x in [p, *p.parents] if (x / 'rag_lab').exists())
sys.path.insert(0, str(root))

import numpy as np
from rag_lab import embed_texts, l2_normalize
v = embed_texts(['semantic search', 'meaning-based lookup', 'a fried egg'],
                provider='ollama', normalize=False)  # raw, un-normalized
```

```python
def cosine_manual(a, b):
    return (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))

print('search vs lookup:', round(cosine_manual(v[0], v[1]), 3))
print('search vs egg   :', round(cosine_manual(v[0], v[2]), 3))
```

Normalize each vector to unit length. Then the denominator is 1, so cosine similarity equals the raw dot product — verify the two match.

```python
u = l2_normalize(v)
print('dot after norm  :', round(float(u[0] @ u[1]), 3))
print('cosine (manual) :', round(cosine_manual(v[0], v[1]), 3))
print('all unit length :', np.allclose(np.linalg.norm(u, axis=1), 1.0))
```

**Takeaway:** normalize once at index time, then similarity is a dot product and top-k is a single `matrix @ query`. That is the engine under every store here.
