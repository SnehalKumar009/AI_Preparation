# 19 · Bridging MCP tools into the agent loop

Phase 3's `Agent` runs think→act→observe over a `ToolRegistry`. The bridge
turns MCP-discovered tools into that registry, so the **same** agent now acts
through MCP servers — no agent changes required.

> Needs a tool-capable model (Ollama `qwen3-coder`, or a cloud key).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
from study_buddy import get_provider
PROVIDER = 'ollama'  # swap: openai | deepseek | anthropic | oxalpha | ollama
llm = get_provider(PROVIDER)
print('using provider:', PROVIDER)
```

```python
from mcp_lab.agent_bridge import build_agent
with build_agent(['filesystem','calculator'], provider=PROVIDER, max_steps=4) as agent:
    res = agent.run('What is 12*9? Then list the files in the mcp_lab directory.')
    print(res.answer)
    print('tool calls:', res.n_tool_calls)
```
