# 03 · Transports: stdio vs HTTP/SSE

MCP messages are transport-agnostic JSON. The two common carriers:

- **stdio** — the host launches the server as a subprocess and reads/writes
  its stdin/stdout. Simplest for *local* servers; used throughout this lab.
- **HTTP + SSE** — the server is a network endpoint; good for remote/shared
  servers. Same messages, different pipe.

Our stdio transport is **newline-delimited JSON**: one message per line.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # PHASE_4_MCP root
import mcp_lab
print('mcp_lab loaded; protocol', mcp_lab.PROTOCOL_VERSION)
```

```python
import io
from mcp_lab.transport import StdioTransport

# Fake a bidirectional pipe with two buffers to show framing:
buf = io.StringIO()
t = StdioTransport(reader=io.StringIO('{"jsonrpc":"2.0","id":1,"method":"ping"}\n'), writer=buf)
print('received:', t.receive())
t.send({'jsonrpc':'2.0','id':1,'result':{}})
print('wire bytes:', repr(buf.getvalue()))
```

Notice each message is a single line terminated by `\n`. `stderr` is left
free for the server's own logging so it never corrupts the protocol stream.
